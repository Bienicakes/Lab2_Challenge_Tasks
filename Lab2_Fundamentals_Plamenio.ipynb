```python
import random
import string

# Fill in your details
SURNAME = "PLAMENIO"  
SEED_NUM = 1154  

# ==========================================
# EXERCISE 1: SENSOR DATA PROCESSING
# ==========================================
print("==========================================")
print("EXERCISE 1: SENSOR DATA PROCESSING")
print("==========================================\n")

random.seed(sum(ord(c) for c in SURNAME.upper()) + SEED_NUM)

raw_pool = [
    "25.4",
    "82.1",
    "105.0",
    "-15.2",
    "45.0",
    "ERR_SENSOR",
    "92.3",
    "INVALID",
    "12.5",
    "68.7",
]
generated_sensor_data = [random.choice(raw_pool) for _ in range(8)]

valid_invalid_results = []
classification_results = []
execution_log_1 = []

for idx, item in enumerate(generated_sensor_data, 1):
    try:
        val = float(item)
        if 0.0 <= val <= 100.0:
            valid_invalid_results.append(f"{item}: VALID")
            if val < 30.0:
                cls = "Low"
            elif val <= 75.0:
                cls = "Optimal"
            else:
                cls = "High"
            classification_results.append(f"{val} -> {cls}")
            execution_log_1.append(
                f"Item #{idx}: [{item}] Passed validation. Classified as {cls}."
            )
        else:
            valid_invalid_results.append(f"{item}: INVALID (Out of Bounds)")
            execution_log_1.append(
                f"Item #{idx}: [{item}] Failed range check (0-100)."
            )
    except ValueError:
        valid_invalid_results.append(f"{item}: INVALID (Non-numeric)")
        execution_log_1.append(
            f"Item #{idx}: [{item}] Non-numeric input encountered."
        )

print(f"Generated Sensor Data: {generated_sensor_data}")
print(f"Valid/Invalid Results: {valid_invalid_results}")
print(f"Classification Results: {classification_results}")
print("\nExecution Log:")
for log in execution_log_1:
    print(f"  {log}")

print(
    f"\nFinal Output:\nTotal Readings: {len(generated_sensor_data)} | Valid: {len(classification_results)} | Invalid: {len(generated_sensor_data) - len(classification_results)}"
)


# ==========================================
# EXERCISE 2: SIGNAL ANALYSIS
# ==========================================
print("\n\n==========================================")
print("EXERCISE 2: SIGNAL ANALYSIS")
print("==========================================\n")

random.seed(sum(ord(c) for c in SURNAME.upper()) + SEED_NUM + 10)

chars = string.ascii_letters + string.digits + " !@#"
gen_signal = "".join(random.choice(chars) for _ in range(10))
proc_signal = gen_signal.strip().upper()

char_analysis = []
letters = digits = specials = 0

for c in proc_signal:
    if c.isalpha():
        char_analysis.append(f"'{c}' -> Letter")
        letters += 1
    elif c.isdigit():
        char_analysis.append(f"'{c}' -> Digit")
        digits += 1
    else:
        char_analysis.append(f"'{c}' -> Special")
        specials += 1

if letters > digits and letters > specials:
    signal_cls = "Text-Dominant"
elif digits > letters and digits > specials:
    signal_cls = "Numeric-Dominant"
else:
    signal_cls = "Mixed Signal"

print(f"Generated Signal: {gen_signal}")
print(f"Processed Signal: {proc_signal}")
print(f"Character Analysis: {char_analysis}")
print(f"Signal Classification: {signal_cls}")
print(
    "\nExecution Log:\n  [LOG] Signal generated, stripped, and converted to uppercase.\n  [LOG] Character-by-character classification performed."
)
print(
    f"\nFinal Output:\nSignal '{proc_signal}' successfully analyzed and classified as {signal_cls}."
)


# ==========================================
# EXERCISE 3: AUTHENTICATION SYSTEM
# ==========================================
print("\n\n==========================================")
print("EXERCISE 3: AUTHENTICATION SYSTEM")
print("==========================================\n")

gen_password = f"{SURNAME.lower()[:3]}{SEED_NUM}"
attempt_limit = 3
attempts = ["1234", "wrongpass", gen_password]

attempts_made = 0
access_result = "DENIED"
sys_state = "LOCKED"
exec_log_3 = []

for entry in attempts:
    attempts_made += 1
    exec_log_3.append(f"Attempt #{attempts_made}: Input '{entry}'")
    if entry == gen_password:
        access_result = "SUCCESS"
        sys_state = "UNLOCKED"
        exec_log_3.append("Password matched. System unlocked.")
        break
    else:
        exec_log_3.append("Incorrect password.")

print(f"Generated Password: {gen_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {access_result}")
print(f"Final System State: {sys_state}")
print("\nExecution Log:")
for log in exec_log_3:
    print(f"  {log}")

print(
    f"\nFinal Output:\nAuthentication completed in {attempts_made}/{attempt_limit} attempts. Access: {access_result} | System State: {sys_state}"
)
```
